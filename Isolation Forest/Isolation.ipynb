```python
import pandas as pd
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np
```

```python
X, _ = make_blobs(n_samples=300, centers=1, cluster_std=1.0, random_state=42, n_features=2)
```

```python
# visualize the data
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title("Data Points")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
# create the Isolation Forest model
model = IsolationForest(contamination=0.1, random_state=42)
# fit the model to the data
model.fit(X)
```

```python
# Predict anomalies
anomaly_scores = model.decision_function(X)
anomalies = model.predict(X)
anomalies
```

```python
# Fetching the indices of anomalies
anomaly_indices = np.where(anomalies == -1)[0]
print("Indices of Anomalies:", anomaly_indices)
```

```python
# visualize the data
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.scatter(X[anomaly_indices, 0], X[anomaly_indices, 1], color='red', s=100, label='Anomalies')
plt.title("Data Points")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()  
```
