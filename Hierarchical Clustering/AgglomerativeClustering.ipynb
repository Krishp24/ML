```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Import dataset
from sklearn.datasets import load_iris
df = pd.DataFrame(load_iris().data)
```

```python
df.columns = load_iris().feature_names
```

```python
df.head()
```

```python
# Scale the data using standard scalar
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
X_scaled = scalar.fit_transform(df)
```

```python
# Applt PCA to reduce dimensions to 2D
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_scaled_imp = pca.fit_transform(X_scaled)
```

```python
pca.explained_variance_
```

```python
# Let's visualize the data
plt.scatter(X_scaled_imp[:,0], X_scaled_imp[:,1], c = load_iris().target)
plt.title('PCA of Iris Dataset')
plt.show()
```

```python
# Agglomerative Clustering
import scipy.cluster.hierarchy as sh

#plot dendrogram
plt.figure(figsize=(10, 7))
plt.ylabel('Distance')
plt.title('Dendrogram')
dendrogram = sh.dendrogram(sh.linkage(X_scaled_imp, method='ward')) 
```

```python
from sklearn.cluster import AgglomerativeClustering
agg = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
agg.fit(X_scaled_imp)
    
```

```python
agg.labels_
```

```python
# Let's visualize the data
plt.scatter(X_scaled_imp[:,0], X_scaled_imp[:,1], c = agg.labels_)
plt.title('PCA of Iris Dataset')
plt.show()
```
