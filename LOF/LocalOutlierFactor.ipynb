```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Generate sample data
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=300, centers=1, cluster_std=0.1, random_state=43)
```

```python
# Visualize the data
plt.scatter(X[:,0], X[:,1], color='blue', s=50)
plt.title('Sample Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
# Local Outlier Factor
from sklearn.neighbors import LocalOutlierFactor

lof = LocalOutlierFactor(n_neighbors=20, algorithm='auto', contamination='auto', n_jobs=-1)
```

```python
outliers = np.where(lof.fit_predict(X)==-1)[0]
```

```python
# visualize the outliers
plt.scatter(X[:,0], X[:,1], color='blue', s=50, label='Data Points')
plt.scatter(X[outliers,0], X[outliers,1], c = 'red')
```
